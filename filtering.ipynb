filtering = refers to the process of selecting specific elements from an array based on a condition. 

```python
import numpy as np
ages = np.array([[25,18,17, 30, 35, 40], [39, 45,15,18,17, 50, ]])
teenagers = ages[ages < 20]
print(teenagers)
adults = ages[(ages >= 20) & (ages < 40)]
print(adults)
evens = ages[ages % 2 == 0]
print(evens)
odds = ages[ages % 2 != 0]
print(odds)
```

```python
adults = np.where((ages >= 20) & (ages < 40), ages, 0)
print(adults)
```
