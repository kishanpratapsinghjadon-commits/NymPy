# Scaler Arithmatic

```python
import numpy as np
a = np.array([1, 2, 3])
a = a/2
print(a)
```

```python
a= a**2
print(a)
```

# Vectorised math func 

```python
radii = np.array([1, 2, 3])
print(np.pi*radii**2)
```

# Element wise arithmetic

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
print(array1 + array2)
print(array1 * array2)
print(array1 / array2)
```

# comparison operators


```python
scores = np.array([90, 80, 70])
print(scores > 80)
print(scores==100)
print(scores[scores > 80])
```
