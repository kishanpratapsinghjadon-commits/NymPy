```python
import numpy as np
rng = np.random.default_rng()
print(rng.integers(low =0, high =100, size=1))
```

```python
#for same value repeat
rng = np.random.default_rng(seed=1) # if we change the seed value we will get different random numbers
print(rng.integers(low =0, high =100, size=(2, 3)))
```

```python
print(np.random.uniform(low=0, high=11, size=(2, 3)))
```

```python
rng = np.random.default_rng()
fruits = ['apple', 'banana', 'cherry', 'date']
fruit = rng.choice(fruits , size=3)  # randomly select a fruit from the list
print(fruit)
```
